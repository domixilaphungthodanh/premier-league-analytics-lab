Chỉ dùng SQLite memory GIẢ LẬP cho bài ghi; không sửa nguồn thật.

```python
import sqlite3
with sqlite3.connect(':memory:') as practice:
    practice.execute('CREATE TABLE demo(id INTEGER PRIMARY KEY,points INTEGER CHECK(points>=0))')
    try:
        practice.execute('BEGIN IMMEDIATE')
        practice.execute('INSERT INTO demo VALUES(1,3)')
        practice.execute('INSERT INTO demo VALUES(2,-1)')
        practice.commit()
    except sqlite3.IntegrityError:
        practice.rollback()
    assert practice.execute('SELECT COUNT(*) FROM demo').fetchone()[0]==0
print('Rollback: không lưu dòng đầu khi dòng sau lỗi')
```
