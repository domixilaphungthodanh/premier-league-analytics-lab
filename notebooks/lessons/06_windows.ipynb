# CTE, ROW_NUMBER và LAG

Dữ liệu thật Ngoại hạng Anh 2025–2026. SQL chạy qua Python.

**Câu hỏi:** Năm trận gần nhất có phải năm vòng gần nhất không?

**Dự đoán:** trước khi chạy, viết mỗi dòng kết quả đại diện cho gì.

```python
from pathlib import Path
import os, sys
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'epl_lab').is_dir())
sys.path.insert(0, str(root))
from epl_lab.learning import database, read_sql
db = database()
print('SQLite chỉ đọc:', db)
```

Lịch thực tế có đá bù; sắp theo ngày/giờ/mã để xác định trận gần nhất.

```python
from epl_lab.analysis import team_cte,scope
read_sql(team_cte()+', numbered AS (SELECT *,ROW_NUMBER() OVER(PARTITION BY team_id ORDER BY played_on DESC,kickoff_time DESC,match_id DESC) rn FROM team_matches) SELECT * FROM numbered WHERE rn<=5 ORDER BY team_id,rn',{'as_of':scope(db)})
```

Tính LAG trên lịch sử trước khi lọc phần trình bày.

```python
read_sql(team_cte()+'SELECT team_id,played_on,points,LAG(points) OVER(PARTITION BY team_id ORDER BY played_on,kickoff_time,match_id) previous_points FROM team_matches ORDER BY team_id,played_on LIMIT 12',{'as_of':scope(db)})
```
